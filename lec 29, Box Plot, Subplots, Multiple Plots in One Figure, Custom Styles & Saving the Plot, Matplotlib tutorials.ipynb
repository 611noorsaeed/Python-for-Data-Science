# 1  Box plot

gives five number of summary



lower value,

max value, mean,

1st perctile (25%), 

3rd perctile (75%), 

ouputliers (dur daraza elaqon mai parhy data points)

```python
import numpy as np 
import matplotlib.pyplot as plt
data = [np.random.normal(0, std, 100) for std in range(1, 4)]
```

```python
plt.boxplot(data)
plt.show()
```

```python

plt.boxplot(data, vert=True, patch_artist=True)
plt.title('Box Plot')
plt.xlabel('Dataset')
plt.ylabel('Value')
plt.show()
```

# 2 Subplots,

```python
# Subplots
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.subplot(2, 1, 1)
plt.plot(x, y1, 'r')
plt.title('Sin(x)')

plt.subplot(2, 1, 2)
plt.plot(x, y2, 'b')
plt.title('Cos(x)')

plt.tight_layout()
plt.show()
```

# Multiple Plots in One Figure

```python
import matplotlib.pyplot as plt

x1 = [1, 2, 3, 4, 5]
y1 = [2, 3, 10, 1, 20]

x2 = [10, 20, 30, 40, 50]
y2 = [15, 40, 15, 100, 10]  

plt.plot(x1, y1, label='X1, Y1', color='blue', marker='o')
plt.plot(x2, y2, label='X2, Y2', color='red', linestyle='--', marker='s')

plt.title('Multiple Functions')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

# Custom Style Plots

```python
# Style and Save
plt.style.use('ggplot')

x = [10,20,30,50,60]
y = [100,30,10,90,40]
plt.plot(x,y, label='sin(x)')

plt.title('Styled Plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.legend()

```

```python
# Save to file
plt.savefig('styled_plot.png', dpi=300)
plt.show()
```
