

# What is a Pie Plot?
A pie plot (or pie chart) is a circular chart divided into slices to show proportions or percentages of a whole.



### When to Use:
To compare parts of a whole (e.g., market share, budget distribution)

When you have categorical data with relative sizes

### X and Y Data Types:
X: Category labels (e.g., 'Apple', 'Samsung')

Y: Numeric values (e.g., sales, percentages) that add up to 100% or a total

##### Only Y-values are plotted; X-values (labels) are used for naming the slices.



#  Pie Chart Usage

```python
import matplotlib.pyplot as plt 


sizes = [15, 30, 45, 10]
labels = ['A', 'B', 'C', 'D']


plt.pie(sizes,labels=labels,autopct="%1.1f%%",shadow=True)
plt.show()
```

# 1. Market Share of Smartphone Brands

```python

labels = ['Samsung', 'Apple', 'Xiaomi', 'Oppo', 'Others']
sizes = [28, 24, 15, 10, 23]
explode = (0.1, 0.1, 0, 0.5, 0)

plt.pie(sizes, labels=labels, explode=explode, autopct='%1.1f%%',shadow=True, startangle=280)
plt.title('Global Smartphone Market Share')
plt.axis('equal')
plt.show()
```

# 2. Budget Allocation for a Project

```python
labels = ['Development', 'Marketing', 'Operations', 'Research', 'Misc']
sizes = [40, 25, 15, 10, 10]
explode = (0.1, 0, 0, 0, 0)

plt.pie(sizes, labels=labels, explode=explode, autopct='%1.1f%%',
        shadow=True, startangle=140)
plt.title('Project Budget Allocation')
plt.axis('equal')
plt.show()
```

# Daily time usage of students

```python
labels = ['Study', 'Sleep', 'Leisure', 'Exercise', 'Others']
sizes = [35, 30, 20, 5, 10]
explode = (0.05, 0.05, 0.05, 0.05, 0.05)

plt.pie(sizes, labels=labels, explode=explode, autopct='%1.1f%%',
        shadow=True, startangle=90)
plt.title("Student's Daily Time Distribution")
plt.axis('equal')
plt.show()
```
