# What is a Histogram?

A histogram is a graphical representation of the distribution of numerical data.
It uses bars to show how many values fall into ranges (bins) of values.

#  Key Points:
Each bar represents a range of values (like 0–10, 10–20, etc.).

The height of the bar shows how many data points fall in that range.

Useful to understand the shape of data: normal, skewed, etc.

    
## X-axis (horizontal):
Type: Numeric values divided into bins (intervals/ranges)

Represents: The range of the data (e.g., ages, income, scores, etc.)

## Y-axis (vertical):
Type: Frequency (counts) — always numeric integers

Represents: How many data points fall into each bin (range on X-axis)

# Histogram Plot

```python
import matplotlib.pyplot as plt 

ages = [5,5,5,20,15,19,10,8,9,10,10,10,10]

plt.hist(ages)
plt.show()
```

# Monthly Salary In USD

```python
# Sample monthly salaries in USD (realistic example)
salaries = [2800, 3200, 4000, 4200, 4500, 4700, 4800, 5000, 5100, 5200,
            5400, 5500, 5700, 6000, 6200, 6300, 6500, 7000, 7200, 7500]

# Plotting histogram
plt.figure(figsize=(10, 6))
plt.hist(salaries, bins=6, color='skyblue', edgecolor='black')

# Titles and labels
plt.title('Employee Salary Distribution (Monthly)', fontsize=16, weight='bold')
plt.xlabel('Salary Range (USD)', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)

# Grid and style
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.box(False)

# Show plot
plt.tight_layout()
plt.show()
```

# Sample student scores out of 100


```python
import matplotlib.pyplot as plt

scores = [45, 50, 55, 60, 62, 65, 67, 68, 70, 72,
          75, 76, 78, 80, 82, 85, 88, 90, 93, 95]

# Plotting histogram
plt.figure(figsize=(10, 6))
plt.hist(scores, bins=[40, 50, 60, 70, 80, 90, 100], color='lightgreen', edgecolor='black')

# Titles and labels
plt.title('Distribution of Exam Scores', fontsize=16, weight='bold')
plt.xlabel('Score Range', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)

# Grid and formatting
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks([40, 50, 60, 70, 80, 90, 100])
plt.yticks(fontsize=10)
plt.box(False)

# Show plot
plt.tight_layout()
plt.show()
```
