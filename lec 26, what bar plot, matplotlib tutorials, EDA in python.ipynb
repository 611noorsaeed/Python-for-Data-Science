#  What is a Bar Plot in Matplotlib?

A bar plot in Matplotlib is used to display categorical data using rectangular bars. The length of each bar represents the value (usually numerical) of that category.

#  When to Use a Bar Plot:

1. To compare values across categories
(e.g., sales per product, population per city)

2. To show frequency or counts of items

3. Ideal when x-axis has distinct labels, not continuous data


# Axes in a Bar Plot:
 
1. X-axis: Categories (e.g., months, products, countries)

2. Y-axis: Values (e.g., counts, revenue, percentage)

```python
import matplotlib.pyplot as plt 

# Bar plot
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]

plt.bar(categories, values, color='skyblue')
plt.title('Bar Plot')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

# Real Example: Product Sales

```python
import matplotlib.pyplot as plt

# Sample data
products = ['Laptop', 'Tablet', 'Phone', 'Monitor', 'Keyboard']
sales = [120, 85, 200, 75, 150]

# Plot
plt.figure(figsize=(8, 5))  # Set figure size
plt.bar(products, sales, color='skyblue', edgecolor='black', width=0.6)

# Customizations
plt.xlabel('Product', fontsize=12)
plt.ylabel('Units Sold', fontsize=12)
plt.title('Monthly Product Sales', fontsize=14)
plt.grid(axis='y', linestyle='--', alpha=0.7)



plt.tight_layout()
plt.show()
```

#  Example: Average Student Scores

```python
import matplotlib.pyplot as plt

# Data
subjects = ['Math', 'Science', 'English', 'History', 'Art']
average_scores = [88, 92, 81, 76, 95]

# Plot
plt.figure(figsize=(8, 5))
bars = plt.bar(subjects, average_scores, color='mediumseagreen', alpha=0.8, edgecolor='darkgreen', linewidth=1.2)

# Titles and labels
plt.title('Average Test Scores by Subject', fontsize=14)
plt.xlabel('Subjects', fontsize=12)
plt.ylabel('Average Score', fontsize=12)
plt.ylim(0, 100)  # Set y-axis range



plt.grid(axis='y', linestyle=':', alpha=0.5)
plt.tight_layout()
plt.show()
```
